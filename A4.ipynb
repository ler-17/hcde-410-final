# Motivation and problem:
The Gutenberg Poetry Corpus is a historical collection of 3 million lines of poetry designed to support creative computational poetic text generation (def: https://arxiv.org/html/2305.08493). As a sociolinguist and poetry enthusiast, I'm interested in the linguistic diversity of the corpus: which language varieties it emphasizes and which it underrepresents. 

Research Questions:
- How are dialect-marked and archaic lexcial features distributed across the Gutenberg Poetry Corpus?
- What does this suggest about which English varieties a generator trained on it would reproduce or omit?
# Data 
I will be analyzing the Gutenberg Poetry Corpus (https://github.com/aparrish/gutenberg-poetry-corpus). This corpus is a subset of the public domain works from Project Gutenberg (https://github.com/aparrish/gutenberg-dammit/) which contains the fulltext versions of the works along with relevant metadata. 

The Poetry Corpus is a JSON formatted dataset with each item containing a line of poetry and an ID linking to the line's source (author, text) in the Gutenberg, Dammit dataset. This dataset includes other important metadata such as the author's birth and death date, primary language, etc. 

Both datasets are free to use, share, and adapt under the MIT license and the creative commmons license (https://creativecommons.org/licenses/by-sa/4.0/). 

19th- and early 20th-century dialect writing may contain racist or other offensive language. My feature searches are likely to surface some of it. I will quote minimum material (if at any), add a content note where offensive lines appear, and try to report counts / patterns over reproducing examples. 
# Unknowns & dependencies:
- The Gutenberg, Dammit metadata includes author birth and death years, but these fields may be missing or inconsistent for some authors.
- The metadata doesn’t reliably include nationality. I’d need to add it from an external source such as Wikidata, which might not be accurate. 
- I can't control how the poetry lines were extracted for the dataset. Line breaks, character recognition errors, and non-poetic text may interfere with data collection.
- The full Gutenberg, Dammit archive is large. Download time, storage, and memory limits on my computer could slow my early work.
- My langauge variation analysis will depend on lists I build myself (although they will be informed by published sources) on archaic English, Scots, other dialect spelling. Whether those lists are adequate is a judgment call and can’t be fully verified in the time available.
