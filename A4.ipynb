# Motivation and problem:
The Gutenberg Poetry Corpus is a historical collection of 3 million lines of poetry designed to support creative computational poetic text generation (def: https://arxiv.org/html/2305.08493). As a sociolinguist and poetry enthusiast, I'm interested in the linguistic diversity of the corpus: which language varieties it emphasizes and which it underrepresents. 

Research Questions:
- How are dialect-marked and archaic lexcial features distributed across the Gutenberg Poetry Corpus?
- What does this suggest about which English varieties a generator trained on it would reproduce or omit?
# Data 
I will be analyzing the Gutenberg Poetry Corpus (https://github.com/aparrish/gutenberg-poetry-corpus). This corpus is a subset of the public domain works of from Project Gutenberg (https://github.com/aparrish/gutenberg-dammit/) which contains the fulltext versions of the works along with relevant metadata. 

The Poetry Corpus is a JSON formatted dataset with each item containing a line of poetry and an ID linking to the line's source (author, text) in the Gutenberg, Dammit dataset. This dataset includes other important metadata such as the author's birth and death date, primary language, etc. 

Both datasets are free to use, share, and adapt under public domain (https://creativecommons.org/licenses/by-sa/4.0/). 
# Unknowns & dependencies:
- Metadata limitation: The Gutenberg, Dammit metadata includes author birth and death years, but these fields may be missing or inconsistent for some authors. I won’t know what share of lines I can place in an era until I join the two datasets in the first days of the project. 
- Author nationality or region: As far as I know, the metadata doesn’t reliably include nationality. If it’s absent, I’d need to add it from an external source such as Wikidata, which depends on how well those records match, or drop it and use era and language tags only. 
- Corpus construction: I can't control how the poetry lines were extracted. Line breaks, character recognition errors, and non-poetic text may interfere with data collection.
- Data access and size. The full Gutenberg, Dammit archive is large. Download time, storage, and memory limits on my machine could slow early work.
